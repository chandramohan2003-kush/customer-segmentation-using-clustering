```python
# Importing dependencies 
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
from sklearn.mixture import GaussianMixture
from sklearn.metrics.pairwise import cosine_distances
```

```python
# Loading the dataset
df = pd.read_csv("D:/Python programming/Projects/Custumer Segmentation/marketing_campaign.csv", sep = "\t")
df.head(5)
```

## Data Preprocessing

```python
# Removing null values and irrelevant columns
cols_to_drop = ['ID', 'Dt_Customer', 'Z_CostContact', 'Z_Revenue']
df.drop(columns=cols_to_drop, inplace=True)
df.dropna(inplace=True)
```

```python
# 1. Create Age from Year_Birth

df["Age"] = 2024 - df["Year_Birth"]

# 2. Calculate total spending across all products

df["Spent"] = (df["MntWines"] + df["MntFruits"] + df["MntMeatProducts"] + df["MntFishProducts"] + df["MntSweetProducts"] + df["MntGoldProds"])
df["Spent"] = np.log1p(df["Spent"])

# 3. Create Living_With feature

df["Living_With"] = df["Marital_Status"].replace({"Married": "Partner","Together": "Partner","Absurd": "Alone","Widow": "Alone",
                                                  "YOLO": "Alone","Divorced": "Alone","Single": "Alone"})

# 4. Calculate total number of children

df["Children"] = df["Kidhome"] + df["Teenhome"]

# 5. Calculate total family size

df["Family_Size"] = (df["Living_With"].replace({"Alone": 1,"Partner": 2})+ df["Children"])
df["Family_Size"] = df["Family_Size"].astype('int')

# 6. Create parenthood indicator

df["Is_Parent"] = np.where(df["Children"] > 0, 1, 0)

# 7. Group Education into broader categories

df["Education"] = df["Education"].replace({"Basic": "Undergraduate","2n Cycle": "Undergraduate","Graduation": "Graduate",
                                           "Master": "Postgraduate","PhD": "Postgraduate"})

# 8. Rename product-spending columns for clarity

df = df.rename(columns={"MntWines": "Wines","MntFruits": "Fruits","MntMeatProducts": "Meat","MntFishProducts": "Fish",
                        "MntSweetProducts": "Sweets","MntGoldProds": "Gold"})


# --------------------------------------------------
# 9. Drop redundant/original features
# --------------------------------------------------

to_drop = ["Marital_Status","Year_Birth",'AcceptedCmp1','AcceptedCmp2','AcceptedCmp3','AcceptedCmp4','AcceptedCmp5']

df.drop(columns=to_drop, inplace=True)
```

## Encoding

```python
le = LabelEncoder()
for col in ['Education', "Living_With"]:
    df[col] = le.fit_transform(df[col])
```

## GMM

```python
bic_scores = []
aic_scores = []
# Try different number of Gaussian components
for k in range(1,21):
    gmm = GaussianMixture(n_components=k,random_state=42)
    gmm.fit(df_pca)
    bic_scores.append(gmm.bic(df_pca))
    aic_scores.append(gmm.aic(df_pca))

# Create comparison DataFrame
gmm_scores = pd.DataFrame({"K": range(1, 21),"BIC": bic_scores,"AIC": aic_scores})
gmm_scores
```

```python
plt.figure(figsize=(8, 5))
plt.plot(gmm_scores["K"],gmm_scores["BIC"],marker="o",label="BIC")
plt.plot(gmm_scores["K"],gmm_scores["AIC"],marker="o",label="AIC")

plt.xlabel("Number of Components (K)")
plt.ylabel("Score")
plt.title("GMM Model Selection using BIC and AIC")

plt.legend()
plt.grid()
plt.show()
```

```python
silhouette_scores = []
for k in range(2, 21):
    # GMM model
    gmm = GaussianMixture(n_components=k,random_state=42)
    # Fit GMM and get cluster labels
    labels = gmm.fit_predict(df_pca)
    # Calculate Silhouette Score
    score = silhouette_score(df_pca, labels)
    silhouette_scores.append(score)

# Create DataFrame
gmm_silhouette_df = pd.DataFrame({"K": range(2, 21),"Silhouette_Score": silhouette_scores})
# Highest score first
gmm_silhouette_df = gmm_silhouette_df.sort_values(by="Silhouette_Score",ascending=False).reset_index(drop=True)
gmm_silhouette_df
```

```python
def plot_gmm_3d(df_pca, k):
    model = GaussianMixture(n_components=k,random_state=42,n_init=10)
    labels = model.fit_predict(df_pca)
    fig = plt.figure(figsize=(8, 6))
    ax = fig.add_subplot(111, projection='3d')
    for cluster in range(k):
        ax.scatter(
            df_pca.iloc[labels == cluster, 0],
            df_pca.iloc[labels == cluster, 1],
            df_pca.iloc[labels == cluster, 2],
            label=f"Cluster {cluster}"
        )
    ax.set_xlabel("PC1")
    ax.set_ylabel("PC2")
    ax.set_zlabel("PC3")
    ax.set_title(f"GMM Clustering (K={k})")
    ax.legend()

    plt.show()
```

```python
plot_gmm_3d(df_pca, 2)
plot_gmm_3d(df_pca, 3)
```

## Spherical KMEANS

```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import Normalizer
from sklearn.cluster import KMeans
from sklearn.pipeline import make_pipeline
from sklearn.metrics import silhouette_score

# ============================================================
# 1. HELPER FUNCTION: SPHERICAL / COSINE K-MEANS PIPELINE
# ============================================================
def get_cosine_kmeans_pipeline(n_clusters, random_state=42):
    """
    L2 Normalization converts Euclidean distance into Cosine distance.
    Formula: d_euclidean^2 = 2 - 2 * cos(theta)
    """
    return make_pipeline(
        Normalizer(norm='l2'),
        KMeans(n_clusters=n_clusters, init='k-means++', n_init=10, max_iter=300, random_state=random_state)
    )

# ============================================================
# 2. COSINE K-MEANS — 2 CLUSTERS
# ============================================================
pipeline_2 = get_cosine_kmeans_pipeline(n_clusters=2)
labels_2 = pipeline_2.fit_predict(df_pca)

# L2 normalize data before evaluating cosine silhouette score
X_norm = Normalizer(norm='l2').fit_transform(df_pca)
silhouette_2 = silhouette_score(X_norm, labels_2, metric='cosine')

print("Silhouette Score for 2 clusters (Cosine):", silhouette_2)

# ============================================================
# 3. COSINE K-MEANS — 3 CLUSTERS
# ============================================================
pipeline_3 = get_cosine_kmeans_pipeline(n_clusters=3)
labels_3 = pipeline_3.fit_predict(df_pca)

silhouette_3 = silhouette_score(X_norm, labels_3, metric='cosine')

print("Silhouette Score for 3 clusters (Cosine):", silhouette_3)
```

```python
import matplotlib.pyplot as plt
from sklearn.preprocessing import Normalizer
from sklearn.cluster import KMeans

def plot_cosine_kmeans_3d(df_pca, k):
    # L2 normalization for cosine-based clustering
    X_norm = Normalizer(norm='l2').fit_transform(df_pca)
    # K-Means
    kmeans = KMeans(n_clusters=k, init='k-means++', random_state=42, n_init=10)
    labels = kmeans.fit_predict(X_norm)
    # 3D plot
    fig = plt.figure(figsize=(8, 6))
    ax = fig.add_subplot(111, projection='3d')

    for cluster in range(k):
        mask = labels == cluster
        ax.scatter(X_norm[mask, 0], X_norm[mask, 1], X_norm[mask, 2], label=f"Cluster {cluster}")
    ax.set_xlabel("PC1")
    ax.set_ylabel("PC2")
    ax.set_zlabel("PC3")
    ax.set_title(f"Spherical K-Means Clustering (K={k})")
    ax.legend()
    plt.show()
```

```python
df_clustered = df.copy()

# To assign cluster label to every observation
df_clustered["Cluster"] = labels_2

# Reloading original dataset only to recover the original Customer ID.
# The ID was dropped earlier from df, but the original row index was retained.
original_df = pd.read_csv("D:/Python programming/Projects/Custumer Segmentation/marketing_campaign.csv",sep="\t")
df_clustered["ID"] = original_df.loc[df.index, "ID"].values

# Rearranging the columns
cols = ["ID", "Cluster"] + [col for col in df_clustered.columns if col not in ["ID", "Cluster"]]
df_clustered = df_clustered[cols]

print("\nFirst 5 customers with their cluster:")
display(df_clustered[["ID", "Cluster"]].head(5))

```

```python
# ============================================================
# BASIC STATISTICS FOR EACH CLUSTER
# ============================================================

cluster_summary = pd.DataFrame({"Customers": df_clustered.groupby("Cluster").size(),
                                "Percentage": df_clustered["Cluster"].value_counts(normalize=True).sort_index() * 100})
cluster_summary["Percentage"] = cluster_summary["Percentage"].round(2)
display(cluster_summary)

# Visualizing cluster sizes
plt.figure(figsize=(5, 5))
ax = sns.countplot(data=df_clustered,x="Cluster")

plt.title("Number of Customers in Each Cluster")
plt.xlabel("Cluster")
plt.ylabel("Number of Customers")
total = len(df_clustered)
for p in ax.patches:
    count = int(p.get_height())
    percentage = 100 * count / total
    ax.annotate(f"{count}\n({percentage:.1f}%)", (p.get_x() + p.get_width() / 2, p.get_height()), ha="center",va="bottom")
plt.tight_layout()
plt.show()
```

```python
# ============================================================
# NUMERICAL PROFILE OF THE TWO CLUSTERS
# ============================================================

# Variables that are most useful for understanding customers
profile_features = ["Income","Age","Recency","Wines","Fruits","Meat","Fish","Sweets","Gold","Spent","NumDealsPurchases",
                    "NumWebPurchases","NumCatalogPurchases","NumStorePurchases","NumWebVisitsMonth","Children","Family_Size"]

# Mean value of each feature within each cluster
cluster_means = df_clustered.groupby("Cluster")[profile_features].mean().T
# Compare clusters using percentage difference
# This tells us how much higher/lower Cluster 1 is compared with Cluster 0.
comparison = cluster_means.copy()
comparison["Difference"] = (comparison[1] - comparison[0])
comparison["% Difference"] = ((comparison[1] - comparison[0])/ comparison[0].replace(0, np.nan)) * 100
display(comparison.round(2))
```

```python
# ============================================================
# STANDARDIZED CLUSTER PROFILE
# ============================================================

# Calculate overall mean and standard deviation
overall_mean = df_clustered[profile_features].mean()
overall_std = df_clustered[profile_features].std()

# Convert cluster means into standardized deviations
# Positive value = above overall average
# Negative value = below overall average
cluster_z = (cluster_means.subtract(overall_mean, axis=0).divide(overall_std, axis=0))
cluster_z["Absolute_Difference"] = (cluster_z[1] - cluster_z[0]).abs()
cluster_z = cluster_z.sort_values("Absolute_Difference",ascending=False)

plt.figure(figsize=(8, 7))
sns.heatmap(cluster_z,annot=True,fmt=".2f",center=0,cmap="RdBu_r")
plt.title("Standardized Customer Profile by Cluster")
plt.xlabel("Cluster")
plt.ylabel("Feature")

plt.tight_layout()
plt.show()
```
